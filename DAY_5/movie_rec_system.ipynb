```python
movies = pd.read_csv(r"C:\Users\AadharJain\Desktop\IBM_Training-main\DAY_5\tmdb_5000_movies.csv")
credits = pd.read_csv(r"C:\Users\AadharJain\Desktop\IBM_Training-main\DAY_5\tmdb_5000_credits.csv")
```

```python
movies = movies.merge(credits, on='title')
movies.shape
```

```python
movies.columns
```

```python
movies = movies[['movie_id', 'title', 'overview', 'genres', 'keywords', 'cast', 'crew']]
movies.head()
```

```python
movies.isnull().sum()
```

```python
movies.dropna(inplace=True)
movies.isnull().sum()
```

```python
import ast
def convert(text):
    L = []
    for i in ast.literal_eval(text):
        L.append(i['name'])
    return L
```

```python
movies['genres'] = movies['genres'].apply(convert)
movies['keywords'] = movies['keywords'].apply(convert)
movies.head()
```

```python
def convert3(text):
    L = []
    cnt = 0
    for i in ast.literal_eval(text):
        if cnt <= 3:
            L.append(i['name'])
        cnt+=1
    return L
```

```python
movies['cast'] = movies['cast'].apply(convert3)
movies.head()
```

```python
def fetch_director(text):
    L = []
    for i in ast.literal_eval(text):
        if i['job'] == 'Director':
            L.append(i['name'])
    return L
```

```python
movies['crew'] = movies['crew'].apply(fetch_director)
movies.head()
```

```python
def collapse(L):
    L1 = []
    for i in L:
        L1.append(i.replace(" ", ""))
    return L1
```

```python
movies['cast'] = movies['cast'].apply(collapse)
movies['crew'] = movies['crew'].apply(collapse)
movies['genres'] = movies['genres'].apply(collapse)
movies['keywords'] = movies['keywords'].apply(collapse)
```

```python
movies.head()
```

```python
movies['overview'] = movies['overview'].apply(lambda x: x.split() if isinstance(x,str) else [])
```

```python
movies.sample(5)
```

```python
movies['tags'] = movies['overview'] + movies['genres'] + movies['keywords'] + movies['cast'] + movies['crew']
movies.head()
```

```python
cleaned_movies = movies[['movie_id', 'title', 'tags']]
cleaned_movies.head()
```

```python
cleaned_movies['tags'] = cleaned_movies['tags'].apply(lambda x: " ".join(x))
cleaned_movies.head()
```

```python
all_movies = cleaned_movies['title']
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=5000, stop_words='english')
```

```python
cv
```

```python
vector = cv.fit_transform(cleaned_movies['tags']).toarray()
```

```python
vector.shape
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vector)
```

```python
similarity
```

```python
import pickle

pickle.dump(cleaned_movies, open("cleaned_movies.pkl", "wb"))
pickle.dump(similarity, open("similarity.pkl", "wb"))
pickle.dump(all_movies, open("all_movies.pkl", "wb"))
```

```python
def recommend(movie):
    if movie not in cleaned_movies['title'].values:
        print("Movie not found")
        return

    index = cleaned_movies[cleaned_movies['title'] == movie].index[0]
    distances = sorted(list(enumerate(similarity[index])), reverse=True, key=lambda x:x[1])
    for i in distances[1:6]:
        print(cleaned_movies.iloc[i[0]].title)
```

```python
recommend('Avatar')
```
